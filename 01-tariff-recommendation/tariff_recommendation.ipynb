## Обработка данных

```python
import pandas as pd
df = pd.read_csv('data/users_behavior.csv')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score 
```

```python
df.info()
```

```python
df.describe()
```

```python
df['is_ultra'].value_counts(normalize=True)
```

```python
df_train, df_valid = train_test_split(df, test_size = 0.2, random_state =12345, stratify = df['is_ultra'])
```

```python
df_train, df_test  = train_test_split(df_train, test_size = 0.25, random_state = 12345, stratify =  df_train['is_ultra'])
```

```python
df_train.shape
```

```python
df_test.shape
```

```python
df_valid.shape
```

```python
features = df_train.drop('is_ultra', axis = 1)
target = df_train['is_ultra']
```

```python
features_valid = df_valid.drop('is_ultra', axis = 1)
target_valid = df_valid['is_ultra']
```

```python
from sklearn.dummy import DummyClassifier
dummy = DummyClassifier(strategy='most_frequent').fit(features, target)
baseline = accuracy_score(target_valid, dummy.predict(features_valid))
baseline
```

Перекос 69/31 означает, что accuracy сама по себе неинформативна и любую модель надо сравнивать с константой 0,694

## Тест Моделей

### Деревья

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
best_depth = 0
best_result  = 0
for depth in range(1, 11):
    tree = DecisionTreeClassifier(random_state=12345, max_depth=depth) 
    tree.fit(features, target) 
    predictions = tree.predict(features_valid) 
    result = accuracy_score(target_valid, predictions) 
    if result > best_result:
        best_depth = depth
        best_result = result
         
    print(f"max_depth: {depth}, accuracy: {result}")
print()
print(f"best_depth: {best_depth}, accuracy: {best_result}")
```

### Леса

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
best_n = 0
best_resualt_2 = 0
best_depth_2 = 0
for n in [10, 20, 30, 40, 50]:
    for depth_2 in range(1, 11):
        forest = RandomForestClassifier(random_state=12345, n_estimators=n, max_depth=depth_2) 
        forest.fit(features, target) 
        predictions_valid_forest = forest.predict(features_valid) 
        result_2 =  accuracy_score(target_valid, predictions_valid_forest)
        if result_2 > best_resualt_2:
            best_resualt_2 = result_2
            best_n = n
            best_depth_2 = depth_2
            best_resualt_2 = result_2
        print(f"depth: {depth_2}, n: {n}, accuracy: {result_2}")   
print(f"best_depth: {best_depth_2}")
print(f"best_n: {best_n}")    
print(f"best_resualt: {best_resualt_2}")    
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state=12345, solver='lbfgs', max_iter=1000) 
model.fit(features, target) 
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
pipe = make_pipeline(StandardScaler(), LogisticRegression(random_state=12345))
```

```python
pipe.fit(features, target) 
```

```python
predictions_pipe = pipe.predict(features_valid)
accuracy_pipe = accuracy_score(target_valid, predictions_pipe)

predictions_logistic = model.predict(features_valid)
accuracy_logistic = accuracy_score(target_valid, predictions_logistic)

print(f'accuracy_pipe: {accuracy_pipe}')
print(f'accuracy_logistic: {accuracy_logistic}')
```

 Несколько конфигураций дали неразличимый результат и выбрана самая экономная.
 Более точной моделью оказался случайный лес с глубиной 6 и 30 деревьями

## Общий вывод

**Задача.** Оператор «Мегалайн» хочет автоматически предлагать клиентам
с архивных тарифов один из двух актуальных — «Смарт» или «Ультра».
В распоряжении было поведение 3214 клиентов за месяц: количество звонков,
суммарные минуты, число сообщений и израсходованный трафик. Нужна модель
классификации с долей правильных ответов не ниже 0,75.

**Как проверялось качество.** Данные разделены на три части в пропорции
60/20/20 со стратификацией по целевому признаку. Гиперпараметры всех моделей
подбирались на валидационной выборке; тестовая выборка не использовалась
до самого конца и была задействована ровно один раз — для замера итоговой
модели. Финальная модель переобучена на обучающей и валидационной выборках
вместе.

**Точка отсчёта.** Классы распределены неравномерно: 69% клиентов на «Смарт»,
31% на «Ультра». Поэтому доля правильных ответов сама по себе неинформативна:
модель, которая всем без разбора советует «Смарт», получает 0,694, не
анализируя ничего. Все результаты ниже сравниваются именно с этим уровнем.

**Результат.** Лучшей оказалась модель случайного леса — 30 деревьев,
максимальная глубина 6. На тестовой выборке она даёт **0,818** против **0,694**
у константного правила, то есть выигрывает **12,4 процентных пункта**.
В переводе на клиентов: из каждых 100 верную рекомендацию получат около 82
против 69 при простейшем правиле — примерно 13 дополнительных клиентов
на каждую сотню. Порог 0,75 пройден с запасом.

Расхождение с валидационной оценкой (0,820) составило 0,2 процентных пункта
при стандартной ошибке около 1,6 — то есть результат воспроизвёлся.

**Сравнение моделей.**

| модель | accuracy на валидации | над константой |
|---|---:|---:|
| случайный лес (30 деревьев, глубина 6) | 0,820 | +12,6 |
| решающее дерево (глубина 3) | 0,799 | +10,6 |
| логистическая регрессия | 0,737 | +4,4 |

Логистическая регрессия фактически провалилась, и это стоит объяснить.
Она отнесла к «Ультра» лишь 6,5% клиентов при реальной доле 31%: из 197
пользователей «Ультра» в валидационной выборке она нашла 35. При этом
из тех, кого она всё же назвала «Ультра», права она в 83% случаев. То есть
модель научилась узнавать только самые очевидные случаи — клиентов
с экстремально высоким трафиком и разговором, — а всех остальных на всякий
случай относит к большинству. Её accuracy 0,737 выглядит приемлемо ровно
потому, что «относить всех к большинству» на таких данных и есть выигрышная
стратегия по этой метрике.

Причина в устройстве модели: логистическая регрессия проводит одну прямую
границу между классами, а «Ультра» и «Смарт» в этих четырёх признаках так
не разделяются. Дерево и лес строят границу из множества условий и потому
улавливают сочетания признаков, недоступные линейной модели.

Несколько конфигураций леса дали статистически неразличимый результат
(разброс в пределах 0,3 п.п. при стандартной ошибке 1,6 п.п.) — из них
выбрана самая экономная по вычислениям.

**Ограничения.**

1. Accuracy не различает типы ошибок. Предложить «Смарт» тому, кому нужен
   «Ультра», и наоборот — разные потери для оператора, а метрика считает их
   одинаковыми. Прежде чем внедрять, стоит договориться с заказчиком о цене
   каждой из двух ошибок и перейти на метрику, которая её учитывает.
2. Признаков всего четыре, и они сильно связаны между собой: звонки и минуты
   описывают почти одно и то же. Потолок качества на таких данных невысок.
3. Различия между близкими конфигурациями лежат в пределах шума — выбор
   «лучшей» модели с точностью до третьего знака здесь не имеет смысла.

**Что дальше.** Главное, чего не хватает для практической пользы, — цены
тарифов и суммы, которую клиент фактически платит. Модель сейчас угадывает,
каким тарифом клиент пользуется, а бизнес-вопрос другой: какой тариф ему
выгоднее предложить. Это разные задачи, и вторая требует данных о деньгах,
а не только о поведении.
